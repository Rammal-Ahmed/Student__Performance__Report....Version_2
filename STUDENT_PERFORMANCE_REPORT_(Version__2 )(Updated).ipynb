```python
# STUDENT PERFORMANCE REPORT ( Version 2 )


print("=" * 45)
print("STUDENT PERFORMANCE REPORT (Version 2)".center(45))
print("=" * 45)
print(


)
student_name = input('What is your Name Mister ? ')

Roll_No = int(input('What is your Roll No. ? '))

chemistry = int(input("Enter your Chemistry obtained marks = "))
maths = int(input("Enter your Maths obtained marks = "))
physics = int(input("Enter your Physics obtained marks = "))
english = int(input("Enter your English obtained marks = "))
science = int(input("Enter your Science obtained marks = "))

total_marks = chemistry + maths + physics + english + science

avg = total_marks / 5

percentage = total_marks / 550 * 100

Highest_Marks = max(chemistry, maths, physics, english, science)
Lowest_Marks = min(chemistry, maths, physics, english, science)

if percentage >= 85:
    grade = 'A'
elif percentage >= 70:
    grade = 'B'
elif percentage >= 60:
    grade = 'C'
elif percentage >= 50:
    grade = 'D'
elif percentage >= 40:
    grade = 'E'
else:
    grade = 'F'

if percentage >= 40:
    result = 'Pass'
else:
    result = 'Fail'
print(



)
print('Student Name = ', student_name)
print('Roll No. = ', Roll_No)
print('Total Marks = ', total_marks)
print('Average Marks = ', avg)
print('Percentage =', f'{percentage:.2f}%')
print('Highest Marks = ', Highest_Marks)
print('Lowest Marks = ', Lowest_Marks)
print('Grade = ', grade)
print('Result = ', result)

print(


)
print("=" * 60)

print("THANK YOU FOR USING OUR STUDENT PERFORMANCE REPORT".center(60))
print("SEE YOU IN THE NEXT REPOSITORIES..!".center(60))
print("=" * 60)
```
